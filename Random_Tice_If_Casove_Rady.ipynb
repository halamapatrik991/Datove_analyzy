# 01. Random, Math

```python
import random
# print (dir(random))
import math
# print ("\n",dir(math))
import matplotlib
# print ("\n",dir(matplotlib))
```

```python
#pseudonahodne cislo z intervalu 0.0 <= x <= 1.0
#komentare ctrl + /
# intellisence ctrl + space
#print(random.__doc__)

print("Nahodne Cislo 0-1 -->",random.random())

#Normalne rozdelenie Gaussova krivka
# Pseudonahodne cislo 0.0 <= x <= 100 vratane
print("\nNahodne cislo z intervalu 0 - 100", random.uniform(0,100))
print("nahodne cele cislo -->", random.randint(1,6)) #nahodne cele cislo random intiger
print("nahodne 3 bytes -->", random.randbytes(3)) #nahodne bytes 

print("\nludolfovo Cislo", math.pi)
print("zaokruhli smerom nahor na najblizsie cele cislo -->", math.ceil(5.0001))
print("zaokruhli smerom nadol na najblizsie cele cislo -->", math.floor(5.9999999))
print("zaokruhlenie klasicke round -->", round(5.4532456)) #toto je zabudovane v pythone

print("\nCele cislo 0-10 -->",math.floor(random.uniform(0,10)))
print("Cele cislo od 0-6 inclusivne vratene 0-6 -->",random.randrange(6))
print("Cele cislo od 1-6 inclusivne vratene 1-6 -->",random.randrange(1,6))

#random generovanie z urcitych moznosti
print("\n",random.choice(["vyhra","prehra","remiza"])) 

#rozdelenie retazca a miesanie
balicek_karty = "eso kral kralovna hornik dolnik".split()

print("pred zamiesanim -->", balicek_karty)
random.shuffle(balicek_karty)
print("po zamiesani -->", balicek_karty)
```

# 03. Casove rady - time series


```python
# XLSX, XLS, XLSM, XLSB
# CSV - ciarkou/bodkociarkou/tabom/medzerou oddelene hodnoty
# TSV - tab separated values - tabom oddelene hodnoty
#pandas - python analysis data lib

import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

data_produkty = pd.read_csv("Produky-ceny.csv")
#print(data_produkty)

#chceme dataframe (df) - tabulku
df_data_produkty = pd.DataFrame(data_produkty,columns = ["Datum","Cena"])

#chceme vlastne indexy cez datumy
df_data_produkty.index = df_data_produkty["Datum"]
del df_data_produkty["Datum"]

df_data_produkty.plot(figsize=(12,8))
plt.ylabel("Cena Uhlia")
plt.xlabel("Datum")
plt.title("Casove Rady")
plt.show()
```
