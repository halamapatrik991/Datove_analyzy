# 03. Pretypovanie(casting)

```python
vyplata = 1956.23
print("vyplata ma hodnutu",vyplata)
print ("je to typ", type(vyplata))

vyplata_cele_cislo = int(vyplata)
print ("\nvyplata_cele_cislo ma hodnotu",vyplata_cele_cislo)
print("je to typ",type(vyplata_cele_cislo))

vyplata_retazec = str(vyplata)
print ("\nvyplata_retazec ma hodnotu",vyplata_cele_cislo)
print("je to typ",type(vyplata_retazec))

#None bude vzdy false lebo je to ako 0, vsetko ostatne je vzdy True, prazdny string "" je false ale string s medzerou " " je true
vyplata_bool = bool(vyplata) 
print ("\nvyplata_bool ma hodnotu",vyplata_bool)
print("je to typ",type(vyplata_bool))

```

# 04. Graf, kniznica numpy (numeric python), matplotlib (math plot library- matematicka vykreslovacia kniznica) 

```python
import numpy
import matplotlib.pyplot as plt #python vykreslovanie - alias as plot
```

```python
#suradnice/koordinaty x,y
#arange = array range 0-10
x = numpy.arange(0,4 * numpy.pi, 0.1) #odkial zaciname, co robime (v tomto pripade pocet kopcov sin), vzorkovanie
y = numpy.sin(x)

#matplotlib.pyplot.plot(x,y) #plot - vykresli
#matplotlib.pyplot.show() #vypise to len output bez kodu a nicoho

plt.title("sinus signal")
plt.xlabel("cas") #label - popis osy
plt.ylabel("napatie")
plt.plot(x,y)
plt.show()


```

# 04. Grafy, zoznamy, polia

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#zoznam / list, heterogenny (hocico) 
zoznam_cisla = [1,2,3,"kupusta",None,True,1690.23]
print(zoznam_cisla)
print(type(zoznam_cisla))

#array / pole, homogenne - bud vsetko cisla, alebo vsetko retazce musi byt 1 typ
y = np.array([35,25,20,20])
menovky = ["notebooky", "tlaciarne", "mysky", "projektory"]
vykrojenie  = [0.1,0,0,0]
tiene = True
# r g b c m y k w
farby_vlastne = ["black", "green", "b", "#CCCCCC"]

print("\n", y)
print(type(y))

plt.title("analyza predaja produkty ABC s.r.o")
plt.pie(y, labels = menovky, startangle = 0, explode = vykrojenie,shadow = tiene, colors = farby_vlastne) #kolacovy graf
plt.legend(title = "produkty ABC s.r.o")
plt.show()

x1 = np.array(["IBM","HP","NESS","Microsoft"])
y1 = np.array([2,5,1,9])


plt.bar(x1,y1)
plt.title("porovnanie firiem")
plt.xlabel("firma")
plt.ylabel("obrat v milionoch")
plt.show()

nahoda = np.random.normal(100,10, 200) # prve okolo akeho cisla sa tocime, v akom rozmedzi +-10, kolko cisel generujeme
plt.hist(nahoda)
plt.show()

```
